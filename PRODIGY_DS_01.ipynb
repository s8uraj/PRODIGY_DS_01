```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())

print("\nGender Distribution:")
print(df["sex"].value_counts())
```

```python

# Create clean age data
age_data = df["age"].dropna()

# Create gender counts
gender_data = df["sex"].value_counts()

print("Cleaned Age Records:", len(age_data))
print("\nGender Distribution:")
print(gender_data)

print("\nMissing Age Values After Cleaning:", age_data.isnull().sum())
```

```python

plt.figure(figsize=(10, 6))

sns.histplot(
    age_data,
    bins=20,
    kde=True,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Titanic Passenger Ages", fontsize=15)
plt.xlabel("Age (Years)")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.savefig("age_distribution.png", dpi=300)
plt.show()
```

```python

gender_counts = df["sex"].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values,
    hue=gender_counts.index,
    palette="pastel",
    legend=False
)

plt.title("Gender Distribution of Titanic Passengers", fontsize=15)
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.savefig("gender_distribution.png", dpi=300)
plt.show()
```

```python

print("===== AGE ANALYSIS =====")

print("Average Age:", round(age_data.mean(), 2))
print("Median Age:", round(age_data.median(), 2))
print("Youngest Passenger:", age_data.min())
print("Oldest Passenger:", age_data.max())

print("\n===== GENDER ANALYSIS =====")

print(df["sex"].value_counts())

print("\nGender Percentage:")
print(
    (df["sex"].value_counts(normalize=True) * 100).round(2)
)
```
